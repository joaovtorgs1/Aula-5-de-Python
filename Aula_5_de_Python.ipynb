# Desafio Sprint 5 - Sistema de Análise de Dados

Projeto desenvolvido em Python para leitura, validação e análise de dados.

O sistema utiliza:

- Manipulação de arquivos CSV
- Expressões regulares (Regex)
- Tratamento de exceções
- Exceção personalizada
- f-strings
- Geração de relatório

```python
import csv
import re
```

```python
class FormatoInvalidoError(Exception):
    pass
```

```python
def validar_email(email):

    padrao = r"^[\w\.-]+@[\w\.-]+\.\w+$"

    return bool(re.match(padrao, email))
```

```python
def validar_cpf(cpf):

    padrao = r"^\d{3}\.\d{3}\.\d{3}-\d{2}$"

    return bool(re.match(padrao, cpf))
```

```python
def validar_telefone(telefone):

    padrao = r"^\(\d{2}\)\s?\d{4,5}-\d{4}$"

    return bool(re.match(padrao, telefone))
```

```python
def validar_data(data):

    padrao = r"^\d{2}/\d{2}/\d{4}$"

    return bool(re.match(padrao, data))
```

```python
dados = """nome,email,cpf,telefone,data
João,joao@gmail.com,123.456.789-00,(98) 99999-1111,15/09/2026
Maria,maria@gmail.com,987.654.321-00,(98) 98888-2222,20/09/2026
Carlos,carlosgmail.com,111.222.333-44,(98) 97777-3333,01/10/2026
Ana,ana@gmail.com,222.333.444-55,98999994444,05/10/2026
Pedro,pedro@gmail.com,333.444.555-66,(98) 95555-6666,10/10/2026
Lucas,lucas@gmail.com,444.555.666-77,(98) 94444-7777,25/10/2026
"""

with open("dados.csv", "w", encoding="utf-8") as arquivo:
    arquivo.write(dados)

print("Arquivo dados.csv criado com sucesso!")
```

```python
def ler_arquivo():

    try:

        with open("dados.csv", "r", encoding="utf-8") as arquivo:

            leitor = csv.DictReader(arquivo)

            dados = list(leitor)

    except FileNotFoundError:

        print("ERRO: O arquivo dados.csv não foi encontrado.")

        return None

    except KeyError:

        print("ERRO: Alguma coluna obrigatória está faltando.")

        return None

    else:

        print("Arquivo lido com sucesso!")

        return dados

    finally:

        print("Finalizando leitura do arquivo...")
```

```python
def analisar_dados(dados):

    validos = []
    invalidos = []

    for pessoa in dados:

        erros = []

        # Validar e-mail
        if not validar_email(pessoa["email"]):
            erros.append("E-mail inválido")

        # Validar CPF
        if not validar_cpf(pessoa["cpf"]):
            erros.append("CPF inválido")

        # Validar telefone
        if not validar_telefone(pessoa["telefone"]):
            erros.append("Telefone inválido")

        # Validar data
        if not validar_data(pessoa["data"]):
            erros.append("Data inválida")

        # Verificar resultado
        if len(erros) == 0:

            validos.append(pessoa)

        else:

            invalidos.append({
                "pessoa": pessoa,
                "erros": erros
            })

    return validos, invalidos
```

```python
def verificar_formato(erros):

    if len(erros) > 0:

        raise FormatoInvalidoError(
            "O registro possui dados inválidos."
        )
```

```python
dados = ler_arquivo()

if dados is not None:

    validos, invalidos = analisar_dados(dados)

    print()
    print("ANÁLISE FINALIZADA")
    print("------------------")
    print(f"Total de registros: {len(dados)}")
    print(f"Registros válidos: {len(validos)}")
    print(f"Registros inválidos: {len(invalidos)}")
```

```python
print("REGISTROS VÁLIDOS")
print("=================")

for pessoa in validos:

    print(
        f"Nome: {pessoa['nome']} | "
        f"E-mail: {pessoa['email']} | "
        f"CPF: {pessoa['cpf']} | "
        f"Telefone: {pessoa['telefone']} | "
        f"Data: {pessoa['data']}"
    )
```

```python
print("REGISTROS INVÁLIDOS")
print("===================")

for item in invalidos:

    pessoa = item["pessoa"]

    print(
        f"Nome: {pessoa['nome']} | "
        f"Erros: {', '.join(item['erros'])}"
    )
```

```python
def gerar_relatorio(validos, invalidos):

    total = len(validos) + len(invalidos)

    percentual = (len(validos) / total) * 100

    with open("relatorio.txt", "w", encoding="utf-8") as arquivo:

        arquivo.write("====================================\n")
        arquivo.write("      RELATÓRIO DE ANÁLISE\n")
        arquivo.write("====================================\n\n")

        arquivo.write(f"Total de registros: {total}\n")
        arquivo.write(f"Registros válidos: {len(validos)}\n")
        arquivo.write(f"Registros inválidos: {len(invalidos)}\n")
        arquivo.write(f"Taxa de aprovação: {percentual:.2f}%\n\n")

        arquivo.write("REGISTROS VÁLIDOS\n")
        arquivo.write("-----------------\n")

        for pessoa in validos:

            arquivo.write(
                f"Nome: {pessoa['nome']} | "
                f"E-mail: {pessoa['email']} | "
                f"CPF: {pessoa['cpf']} | "
                f"Telefone: {pessoa['telefone']} | "
                f"Data: {pessoa['data']}\n"
            )

        arquivo.write("\nREGISTROS INVÁLIDOS\n")
        arquivo.write("-------------------\n")

        for item in invalidos:

            pessoa = item["pessoa"]

            arquivo.write(
                f"Nome: {pessoa['nome']} | "
                f"Erros: {', '.join(item['erros'])}\n"
            )

    print("Relatório criado com sucesso!")
```

```python
gerar_relatorio(validos, invalidos)
```
